```python
# Import libraries
import pandas as pd
import numpy as np

# Load datasets
sales = pd.read_csv("sales_orders_integrated.csv")
purchase = pd.read_csv("purchase_orders_integrated.csv")
stock = pd.read_csv("stock_inventory_integrated.csv")

# Display dataset dimensions
print("Sales dataset:", sales.shape)
print("Purchase dataset:", purchase.shape)
print("Stock dataset:", purchase.shape)
```

```python
# Preview datasets
display(sales.head())
display(purchase.head())
display(stock.head())
```

```python
# Check missing values
print("Sales missing values:")
display(sales.isnull().sum()[sales.isnull().sum() > 0])

print("Purchase missing values:")
display(purchase.isnull().sum()[purchase.isnull().sum() > 0])

print("Stock missing values:")
display(stock.isnull().sum()[stock.isnull().sum() > 0])
```

```python
# Total Sales Revenue

total_sales = sales["line_sales_value"].sum()

print(total_sales)
```

```python
# Total Units Sold

total_units = sales["quantity"].sum()

print(total_units)
```

```python
# Total Sales Orders

total_orders = sales["so_id"].nunique()

print(total_orders)
```

```python
# Average Order Value

average_order_value = total_sales / total_orders

print(average_order_value)
```

```python
kpi_results = pd.DataFrame({
    "KPI": [
        "Total Sales Revenue",
        "Total Units Sold",
        "Total Sales Orders",
        "Average Order Value"
    ],
    "Value": [
        total_sales,
        total_units,
        total_orders,
        average_order_value
    ]
})

kpi_results
```

```python
#Procurement Spend

total_procurement_spend = purchase["line_purchase_value"].sum()

print(total_procurement_spend)
```

```python
#Units Purchased

total_procurement_spend = purchase["line_purchase_value"].sum()

print(total_procurement_spend)
```

```python
#Number of purchase orders

total_purchase_orders = purchase["po_id"].nunique()

print(total_purchase_orders)
```

```python
#Average PO Value

average_po_value = (
    total_procurement_spend / total_purchase_orders
)

print(average_po_value)
```

```python
#Supplier Analysis

supplier_spend = (
    purchase.groupby("supplier_name")["line_purchase_value"]
    .sum()
    .sort_values(ascending=False)
)

supplier_spend
```

```python
import pandas as pd
import numpy as np
```

```python
sales = pd.read_csv("sales_orders_integrated.csv")
purchase = pd.read_csv("purchase_orders_integrated.csv")
inventory = pd.read_csv("stock_inventory_integrated.csv")
products = pd.read_csv("products.csv")
```

```python
print(sales.shape)
print(purchase.shape)
print(inventory.shape)
print(products.shape)
```

```python
sales_product = sales.merge(
    products[["product_id", "product_name", "category"]],
    on="product_id",
    how="left"
)
```

```python
product_sales = (
    sales_product.groupby("product_name")["line_sales_value"]
    .sum()
    .sort_values(ascending=False)
)

product_units = (
    sales_product.groupby("product_name")["quantity"]
    .sum()
    .sort_values(ascending=False)
)
print(product_sales,
     product_units)
```

```python
product_order_frequency = (
    sales_product.groupby("product_name")["so_id"]
    .nunique()
    .sort_values(ascending=False)
)
print(product_order_frequency)
```

```python
product_sales = (
    sales_product.groupby("product_name")["line_sales_value"]
    .sum()
    .sort_values(ascending=False)
)
print(product_sales)
```

```python
product_units = (
    sales_product.groupby("product_name")["quantity"]
    .sum()
    .sort_values(ascending=False)
)
print(product_units)
```

```python
#Branch Analysis

branch_sales = (
    sales.groupby("branch_name")["line_sales_value"]
    .sum()
    .sort_values(ascending=False)
)
print(branch_sales)
```

```python
#Supplier Delivery Analysis

purchase["delivery_status"].value_counts()

average_delay = purchase["delivery_delay_days"].mean()

print(average_delay)
```
