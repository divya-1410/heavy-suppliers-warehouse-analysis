We can also check whether the grand total is mathematically consistent:

```python
df["calculated_grand_total"] = (
    df["total_cost"] + df["total_gst_amount"]
)

mismatch = ~np.isclose(
    df["grand_total"],
    df["calculated_grand_total"],
    atol=0.02
)

print("Grand Total mismatches:", mismatch.sum())
```

grand_total = total_cost + total_gst_amount

is consistent across the dataset.

2.4 Date Consistency

```python
# Convert dates
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
df["expected_delivery_date"] = pd.to_datetime(
    df["expected_delivery_date"], errors="coerce"
)
df["received_date"] = pd.to_datetime(
    df["received_date"], errors="coerce"
)

# Order date should not be after expected delivery
invalid_order_expected = (
    df["order_date"] > df["expected_delivery_date"]
)

# Received date should not be before order date
invalid_received_order = (
    df["received_date"] < df["order_date"]
)

print(
    "Order date > Expected delivery:",
    invalid_order_expected.sum()
)

print(
    "Received date < Order date:",
    invalid_received_order.sum()
)
```

Therefore, there are no violations of these two date rules.

We can also identify late deliveries:

```python
late_delivery = (
    df["received_date"] > df["expected_delivery_date"]
)

print("Late deliveries:", late_delivery.sum())
```

There are 3,267 received purchase orders whose received date is later than the expected delivery date.

2.6 Business Rule Checks

```python
# Rule 1: PO ID should be unique
rule1 = df["po_id"].is_unique

# Rule 2: Financial values should not be negative
rule2 = (df[financial_columns] >= 0).all().all()

# Rule 3: Order date <= expected delivery date
rule3 = (df["order_date"] <= df["expected_delivery_date"]).all()

# Rule 4: Received date >= order date when received date exists
rule4 = (
    df.loc[df["received_date"].notna(), "received_date"]
    >=
    df.loc[df["received_date"].notna(), "order_date"]
).all()

# Rule 5: Grand total = cost + GST
rule5 = np.isclose(
    df["grand_total"],
    df["total_cost"] + df["total_gst_amount"],
    atol=0.02
).all()

print("PO ID unique:", rule1)
print("No negative financial values:", rule2)
print("Order <= Expected Delivery:", rule3)
print("Received >= Order:", rule4)
print("Grand Total calculation valid:", rule5)
```

5. **Data** **Cleaning** / **Preparation**

5.1 Data Type Standardization

```python
# ID columns as strings
id_columns = [
    "po_id",
    "supplier_id",
    "branch_id"
]

for column in id_columns:
    df[column] = df[column].astype("string").str.strip()

# Date columns
date_columns = [
    "order_date",
    "expected_delivery_date",
    "received_date"
]

for column in date_columns:
    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )

# Numeric columns
financial_columns = [
    "total_cost",
    "total_gst_amount",
    "grand_total"
]

for column in financial_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print(df.dtypes)
```

The dataset is standardized so that:

* IDs are string values.
* Dates are datetime values.
* Financial fields are numeric.

5.2 Invalid Value Handling

```python
valid_statuses = ["Received", "Cancelled"]

invalid_status = ~df["po_status"].isin(valid_statuses)

print("Invalid status records:", invalid_status.sum())
```

No invalid status categories were found.

The missing received_date values correspond to cancelled purchase orders, so they should not automatically be replaced with another date.

5.3 Duplicate Handling

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)
```

No duplicate rows were removed.

5.4 Category Standardization

```python
# Remove unnecessary spaces
df["po_status"] = df["po_status"].str.strip()

# Standardize capitalization
df["po_status"] = df["po_status"].str.title()

# Check final categories
print(df["po_status"].value_counts())
```

5.5 Final Validation

```python
print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDuplicate PO IDs:")
print(df["po_id"].duplicated().sum())

print("\nStatuses:")
print(df["po_status"].value_counts())

print("\nNegative financial values:")

for column in financial_columns:
    print(
        column,
        (df[column] < 0).sum()
    )

print("\nDate consistency:")
print(
    "Order > Expected:",
    (df["order_date"] > df["expected_delivery_date"]).sum()
)

print(
    "Received < Order:",
    (
        df.loc[df["received_date"].notna(), "received_date"]
        <
        df.loc[df["received_date"].notna(), "order_date"]
    ).sum()
)

print("\nGrand total mismatches:")

grand_total_check = ~np.isclose(
    df["grand_total"],
    df["total_cost"] + df["total_gst_amount"],
    atol=0.02
)

print(grand_total_check.sum())
```

The final dataset passes the major structural, financial, and date validation checks.
